## 1. Reconstruct the original result

We reload the initial observations so this notebook does not depend on the Python variables from Part 1. We then apply the **same rule**: Group A when the first coordinate is negative, Group B otherwise.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import silhouette_score

columns = ["measurement_1", "measurement_2"]
first = pd.read_csv("../data/exercises/clustering_first.csv")
assert first.columns.tolist() == columns
assert len(first) == 90
assert first[columns].notna().all().all()

first_labels = (first["measurement_1"] >= 0).astype(int)
first_score = silhouette_score(first[columns].to_numpy(), first_labels)
print(f"Original sample silhouette: {first_score:.3f}")
```

## 2. Apply the fixed rule to independent observations

Load 90 fresh synthetic observations. Use the original partition rule **without refitting or tuning**. Compare the geometric arrangement with the original sample.

```python
fresh = pd.read_csv("../data/exercises/clustering_fresh.csv")
assert fresh.columns.tolist() == columns
assert len(fresh) == 90
assert fresh[columns].notna().all().all()

fresh_labels = (fresh["measurement_1"] >= 0).astype(int)

fig, ax = plt.subplots(figsize=(6, 5))
for group, name in [(0, "Group A"), (1, "Group B")]:
    subset = fresh.loc[fresh_labels == group]
    ax.scatter(
        subset["measurement_1"], subset["measurement_2"],
        s=40, alpha=0.85, label=name,
    )
ax.axvline(0, color="black", linestyle="--", linewidth=1, alpha=0.7)
ax.set(
    xlabel="Fresh measurement 1",
    ylabel="Fresh measurement 2",
    title="Independent observations — same fixed rule",
)
ax.set_aspect("equal", adjustable="box")
ax.legend()
plt.show()
```

Calculate the fresh-sample silhouette score and compare it with your prediction. The data differ, even when the two scores happen to be similar.

```python
fresh_score = silhouette_score(fresh[columns].to_numpy(), fresh_labels)

print(f"Original sample silhouette: {first_score:.3f}")
print(f"Fresh sample silhouette:    {fresh_score:.3f}")
print("Fresh group sizes:", fresh_labels.value_counts().sort_index().to_dict())
```

## Optional extension — compare with K-means

If you finish early, fit K-means with two clusters to **the original sample** and compare its silhouette score with the fixed sign rule. Does optimizing a geometric clustering measure establish that two scientifically meaningful populations exist?

```python
from sklearn.cluster import KMeans

X_first = first[columns].to_numpy()
kmeans_labels = KMeans(n_clusters=2, n_init=20, random_state=42).fit_predict(X_first)
kmeans_score = silhouette_score(X_first, kmeans_labels)

print(f"Fixed sign rule on original data: {first_score:.3f}")
print(f"K-means on original data:         {kmeans_score:.3f}")
```
