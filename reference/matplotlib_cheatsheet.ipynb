**Figure vs Axes:** the **Figure** is the whole picture, while an **Axes** object, conventionally named `ax`, contains a plotting area, axes, labels and data. In the examples below, pandas often **returns an Axes** when it creates a chart, and we use `ax` to adjust it.

All claims and temperatures shown here are **synthetic classroom values**, not industry or observed climate statistics.

[Official Matplotlib quick start](https://matplotlib.org/stable/users/explain/quick_start.html).


## 1. Create small datasets

These examples require no downloads and no other notebooks. The data do **not** describe actual establishments, workers or temperature histories.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "sector": ["Transport", "Transport", "Retail", "Retail",
               "Services", "Services", "Transport", "Services"],
    "worker_years": [10., 90., 40., 55., 20., 80., 30., 50.],
    "claims_count": [1, 1, 0, 1, 0, 2, 1, 1],
})

years = np.arange(2016, 2026)
annual = pd.Series(
    [27.2, 27.3, 27.0, 27.4, 27.6, 27.2, 27.5, 27.8, 27.6, 27.9],
    index=years,
    name="invented_temperature_c",
)

data.head()
```

## 2. Histogram — how are counts distributed?

A pandas plotting method can create an Axes object. The axes labels explain the values and **units**. A histogram groups numeric values into bins.


```python
ax = data["claims_count"].plot.hist(
    bins=5,
    figsize=(8, 3),
    title="Invented claims per establishment-year",
)
ax.set_xlabel("Synthetic claim count")
ax.set_ylabel("Number of establishment-years");
```

In JupyterLab with the usual inline plot backend, the figure appears automatically. `plt.show()` is **normally unnecessary** for these simple notebook cells.

The final semicolon (`;`) hides an unhelpful text representation such as `Text(...)`, but **does not hide the figure**. It is optional.


## 3. Bar chart — compare categories

First calculate the sector-level rate **using total claims divided by total worker-years**. Remember that worker-years measure exposure, not necessarily the number of workers.


```python
by_sector = data.groupby("sector")[["claims_count", "worker_years"]].sum()
by_sector["claims_per_100_worker_years"] = (
    100 * by_sector["claims_count"] / by_sector["worker_years"]
)
by_sector.round(2)
```

We create a horizontal bar chart with pandas, then use `ax` to label it.


```python
ax = by_sector["claims_per_100_worker_years"].sort_values().plot.barh(
    figsize=(8, 3),
    title="Invented claim frequencies by sector",
)
ax.set_xlabel("Claims per 100 worker-years")
ax.set_ylabel("Fictional sector");
```

## 4. Scatter plot — how are two quantities related?

Compare exposure and observed synthetic claims. More exposure provides more opportunities for claims, but **a plotted association is not proof of causation**.


```python
ax = data.plot.scatter(
    x="worker_years",
    y="claims_count",
    figsize=(8, 3),
    title="Invented exposure and claim counts",
)
ax.set_xlabel("Worker-years of exposure")
ax.set_ylabel("Synthetic claim count");
```

## 5. Line chart — how does a value change over time?

The next plot is **not NASA POWER data or weather-station observations**; it uses the invented annual values created at the beginning. In the climate project, always identify NASA POWER results as *model-informed grid-cell estimates*.


```python
ax = annual.plot(
    marker="o",
    figsize=(8, 3),
    title="Invented annual temperature series",
)
ax.set_xlabel("Year")
ax.set_ylabel("Synthetic temperature (°C)");
```

## 6. Change titles, labels, limits and grids

These methods operate on an existing `ax`: `set_xlabel`, `set_ylabel`, `set_title`, `set_xlim` and `grid`. You can use them with nearly any chart.

This example makes a **new** scatter chart to keep it independent of earlier plotting cells.


```python
ax = data.plot.scatter(x="worker_years", y="claims_count", figsize=(8, 3))
ax.set_xlabel("Worker-years")
ax.set_ylabel("Synthetic claims")
ax.set_title("Illustration: labels and axes")
ax.set_xlim(left=0)
ax.grid(alpha=0.3);
```

The alternative `ax.set(xlabel="...", ylabel="...", title="...")` can set several properties together. For beginners, separate methods may be easier to read.


## 7. Construct a figure yourself with `plt.subplots()`

For some plots, you need to create the Figure and Axes explicitly. `plt.subplots()` returns both objects. After that, use methods on `ax`.

Here we compare entirely fictional observed claim counts with **expected** counts. Expected values need not be integers, and a model cannot predict every random realized count exactly.


```python
observed = np.array([0, 1, 2, 1, 3, 0, 2, 1])
predicted = np.array([0.6, 1.2, 1.7, 1.5, 2.4, 0.8, 1.9, 1.0])

fig, ax = plt.subplots(figsize=(7, 4))
ax.scatter(observed, predicted, alpha=0.7, label="Fictional establishment-years")
ax.plot([0, 3.2], [0, 3.2], linestyle="--", label="Equal values")
ax.set_xlabel("Observed synthetic claims")
ax.set_ylabel("Predicted mean claims")
ax.legend()
ax.grid(alpha=0.3);
```

`fig` refers to the whole Figure. `ax` is its plot area. We use `ax.legend()` because the labelled series should be distinguished.


## 8. Compare an annual series with a fitted line

This example is only about **plotting**. It fits an illustrative straight line to the invented annual temperatures using NumPy. A scientifically defensible climate-trend analysis needs data-quality checks and uncertainty analysis that are beyond this plotting reference.


```python
slope, intercept = np.polyfit(years, annual.to_numpy(), deg=1)
fitted = slope * years + intercept

ax = annual.plot(marker="o", label="Invented annual temperatures", figsize=(8, 3))
ax.plot(years, fitted, label="Illustrative fitted line")
ax.set_xlabel("Year")
ax.set_ylabel("Synthetic temperature (°C)")
ax.legend();
```

The two series must have compatible units and x-coordinates. A fitted model line is **not** another observed temperature record.


## 9. Optional: two panels when they genuinely help

Use a pair of Axes only when you want to compare two related outputs. Here the right panel shows **residuals**, the difference between the invented annual values and the fitted line.


```python
residuals = annual.to_numpy() - fitted

fig, (ax_left, ax_right) = plt.subplots(1, 2, figsize=(11, 4))

ax_left.plot(years, annual.to_numpy(), marker="o")
ax_left.set_title("Invented temperatures")
ax_left.set_xlabel("Year")
ax_left.set_ylabel("Synthetic temperature (°C)")

ax_right.plot(years, residuals, marker="o")
ax_right.axhline(0, linestyle="--")
ax_right.set_title("Residuals from illustrative fit")
ax_right.set_xlabel("Year")
ax_right.set_ylabel("Residual (°C)")

fig.tight_layout();
```

Use **separate cells and plots** when figures address unrelated questions. A quick reference does not need elaborate graphics to be useful.


## 10. Check whether a plot is misleading

- Are the **units** stated on the axes?
- Does the title say when the data are **synthetic**, modelled or observed?
- Are axis limits and bin widths appropriate?
- Are the plotted series clearly labelled?
- Could a fitted trend be mistaken for an observation?
- Does the chart answer the question you actually want to investigate?
